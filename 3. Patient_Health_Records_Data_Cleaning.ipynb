Convert to datetime

```python
print("Data type:", df_clean["last_visit_date"].dtype)

print("\nMissing dates:", df_clean["last_visit_date"].isna().sum())

print("\nDate range:")
print("Earliest:", df_clean["last_visit_date"].min())
print("Latest:", df_clean["last_visit_date"].max())

print("\nSample:")
print(df_clean["last_visit_date"].head(20))
```

```python
print(df.columns.tolist())
```

```python
df_clean["last_visit_date"] = df["Last_Visit_Date"].copy()
```

```python
date = df_clean["last_visit_date"].astype(str).str.strip()

parsed_date = pd.Series(
    pd.NaT,
    index=df_clean.index,
    dtype="datetime64[ns]"
)

# YYYY-MM-DD
mask_iso = date.str.match(r"^\d{4}-\d{2}-\d{2}$")
parsed_date.loc[mask_iso] = pd.to_datetime(
    date.loc[mask_iso],
    format="%Y-%m-%d",
    errors="coerce"
)

# Apr 15 2021
mask_text = date.str.match(r"^[A-Za-z]{3} \d{2} \d{4}$")
parsed_date.loc[mask_text] = pd.to_datetime(
    date.loc[mask_text],
    format="%b %d %Y",
    errors="coerce"
)

# YYYYMMDD
mask_compact = date.str.match(r"^\d{8}$")
parsed_date.loc[mask_compact] = pd.to_datetime(
    date.loc[mask_compact],
    format="%Y%m%d",
    errors="coerce"
)

# DD/MM/YYYY
mask_slash = date.str.match(r"^\d{2}/\d{2}/\d{4}$")
parsed_date.loc[mask_slash] = pd.to_datetime(
    date.loc[mask_slash],
    format="%d/%m/%Y",
    errors="coerce"
)

df_clean["last_visit_date"] = parsed_date
```

```python
print("Missing dates:", df_clean["last_visit_date"].isna().sum())
print("Data type:", df_clean["last_visit_date"].dtype)
print("Earliest:", df_clean["last_visit_date"].min())
print("Latest:", df_clean["last_visit_date"].max())
```

## 4.14 diagnosis_code

```python
print(df_clean["diagnosis_code"].value_counts(dropna=False).head(30))

print("\nData type:", df_clean["diagnosis_code"].dtype)

print("\nSample values:")
print(df_clean["diagnosis_code"].dropna().head(30).to_string(index=False))
```

```python
# Standardize diagnosis code text
df_clean["diagnosis_code"] = df_clean["diagnosis_code"].str.strip()

# Treat "unknown" as missing/unknown
df_clean["diagnosis_code"] = df_clean["diagnosis_code"].replace(
    {"unknown": np.nan}
)

# Validate diagnosis code format
valid_diagnosis = df_clean["diagnosis_code"].str.match(
    r"^[A-Z]\d{2}(\.\d+)?$",
    na=False
)

# Count invalid diagnosis codes
print("Invalid diagnosis codes:", (~valid_diagnosis & df_clean["diagnosis_code"].notna()).sum())

# Convert invalid codes to NaN
df_clean.loc[
    (~valid_diagnosis) & df_clean["diagnosis_code"].notna(),
    "diagnosis_code"
] = np.nan
```

```python
print(df_clean["diagnosis_code"].value_counts(dropna=False))

print("\nData type:", df_clean["diagnosis_code"].dtype)

print("\nRemaining invalid codes:")
print(
    df_clean.loc[
        df_clean["diagnosis_code"].notna() &
        ~df_clean["diagnosis_code"].str.match(
            r"^[A-Z]\d{2}(\.\d+)?$",
            na=False
        ),
        "diagnosis_code"
    ].unique()
)
```
